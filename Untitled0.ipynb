```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Get the uploaded image filename automatically
filename = next(iter(uploaded.keys()))

# Load image
image_bgr = cv2.imread(filename)

# Check whether image loaded successfully
if image_bgr is None:
    print("Image could not be loaded.")
else:
    # Convert BGR to RGB
    image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

    # Inspect dimensions
    height, width, channels = image.shape

    print("Image dimensions:", image.shape)
    print("Height:", height)
    print("Width:", width)
    print("Channels:", channels)
    print("Data type:", image.dtype)
    print("First pixel RGB values:", image[0, 0])

    # Display original image
    plt.figure(figsize=(7, 5))
    plt.imshow(image)
    plt.title("Original RGB Image")
    plt.axis("off")
    plt.show()
```

```python

# Create a folder for output images
os.makedirs("outputs", exist_ok=True)

# 1. Convert RGB image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

# 2. Apply Gaussian Blur
blurred = cv2.GaussianBlur(image, (15, 15), 0)

# 3. Apply Canny Edge Detection
edges = cv2.Canny(gray, 100, 200)

# Display all four images
plt.figure(figsize=(12, 9))

plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(blurred)
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(edges, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.savefig("outputs/comparison.png", dpi=150)
plt.show()

# Save individual output images
cv2.imwrite("outputs/grayscale.png", gray)
cv2.imwrite(
    "outputs/gaussian_blur.png",
    cv2.cvtColor(blurred, cv2.COLOR_RGB2BGR)
)
cv2.imwrite("outputs/canny_edges.png", edges)

print("All images processed and saved successfully!")
```
